## 自分のアイデアに取り組み始める方法

さて、あなたは何か作りたいもののアイデアを思いついたとします。

おめでとうございます！それこそが最も重要な部分であり、もう終わっているのです。アイデアこそがすべてです！

とはいえ、どこから始めればいいのか分からない、という方も多いでしょう。これについては本当によく質問をもらいます。

そこで、3つの重要なアドバイスをお伝えしたいと思います。


## アドバイス1: 科学者になる

AIエンジニアであることは、2つの帽子をかぶることを意味します。ソフトウェアエンジニアとしての帽子と、データサイエンティストとしての帽子です。

プロジェクトの開始時点では、ソフトウェアエンジニアの帽子を脱ぎ、データサイエンティストの帽子をしっかりとかぶることが極めて重要です。

ソフトウェアエンジニアリングの出身者（私自身も含めて！）の多くは、これに苦労します。人は自分の得意な領域、つまりアーキテクチャ、設計、データパイプライン、スケーラビリティ、デプロイといった質問に走りがちです。

これらはどれも素晴らしい質問ですが、この初期段階において重要な質問ではありません。本当に難しい質問でもありません。あなたのアイデアの成否を決めるものではないのです。

重要な質問に取り組むには、科学者にならなければなりません。

自分自身に問いかけてみましょう。
1. モデルの成功をどのように評価しますか？何を測定しますか？
2. どんなデータを持っていて、どんなデータが必要ですか？
3. さまざまな手法を試し、モデルの性能を理解するための小さなプロトタイプをどのように構築できますか？

まず科学的な問いに取り組みましょう。LLMを使って何が達成できるのか、そしてそれをどう実現するのかです。

そして、Pythonモジュールで作業を始める前に、まずノートブックから始めることをお勧めします。そうすることで、自分自身にこの科学的な考え方を強制することができます。

これはエージェントを使ったプロジェクトにおいては、なおさら重要になります。エージェントたちがどのように協働するかについて、技術図のような大きなエージェントアーキテクチャ図を描きたくなるかもしれません。しかし、これはソフトウェアエンジニアの考え方です。私がお勧めするのは、別のアプローチです。すなわち、さまざまな手法を試し、何がうまく機能するかを調べ、仮説を検証し、そして反復していくことです。

## アドバイス2: R&Dを大切にする

さて、これはアドバイス1とかなり似ていますが、繰り返す価値があります！

私はよくこんな質問を受けます。「Ed、こういうソリューションを作りたいんです。モデルA、B、Cのどれを使うべきでしょうか？Agents、RAG、fine-tuningのどれを使うべきですか？」

私の答えはほぼ常に同じです。
- 上記のすべてをやるべきです！大切なのは実験することです。(a) 評価基準を決める、(b) 精選されたデータセットを用意する、(c) さまざまなアイデアを試し、その性能を見る
- 私自身、どれが一番うまくいくかについて直感を持つことは多いですが、その直感はしばしば間違っています！私を信じずに、自分自身で試してみてください
- 実験に代わるものは、断じてありません。

## アドバイス3: 大きな夢を持て、しかし小さく始めよ！

私はよく、学生から200〜300行のコードを送られて「Ed、これがうまく動きません、直してください」と言われることがあります。😂

デバッグガイドでも書いた通り、それはうまくいく方法ではありません！

プロジェクトを始めるときは、小さく、シンプルに始めることが何より重要です。すべてのプロンプトを完璧に仕上げ、すべてのステップを細部まで丁寧に作り込みましょう。それぞれのLLM呼び出しが望んだとおりに機能していることを自分で確認し、応答が一貫して安定するまで入力を反復して調整してください。

常に段階的に、小さく検証可能な構成要素で作業を進めていけば、何が起きているのかを完全に把握できるはずです。

もちろん、私はプロジェクトのお手伝いをすることが大好きですし、問題の解決を助けるのは素晴らしいことです。ですが、プロジェクトに段階的に取り組んでいれば、どこがうまく機能していないのかを常に明確に把握できるはずです。そうすれば、私が手助けするのも非常に効率的になります。

# まとめ

この3つのアドバイスに共通するテーマにお気づきかもしれません。データサイエンス出身でない人にとって、このような場当たり的なやり方で問題に取り組むのは、どこか落ち着かず、「小手先」で満足感のないものに感じられるかもしれません。しかし私の考えでは、これこそが優れたAIエンジニアになるために身につけるべき、最も重要なスキルなのです。不確実性に対して心地よくいられること、実験を楽しむこと、そして科学者であることを受け入れること。
