favorite_fruitがnotebookのさらに下で変更されたにもかかわらず、表示される内容が変わっているのが分かるでしょうか?

# 「カーネル」の説明

このnotebookの裏側には、セルを実行するたびにそれを実行するPythonのプロセスがあります。このPythonのプロセスはカーネル(Kernel)と呼ばれています。各notebookはそれぞれ独立したカーネルを持っています。

上にある「Restart Kernel」というボタンをクリックしてみてください。

その後、次のセルを実行しようとするとエラーになります。なぜなら、favorite_fruitがもう定義されていないからです。notebookの先頭からセルを再度実行する必要があります。そうすれば、次のセルは問題なく実行できるはずです。

````python
print(f"My favorite fruit is {favorite_fruit}")
````

# セルの追加と削除

このセルをクリックし、上のツールバーにある[+ Code]ボタンをクリックすると、このセルのすぐ下に新しいセルが作成されます。前のセルのコードをコピー&ペーストして、実行してみましょう!選択したセルの右上には、それを削除するためのアイコン(ゴミ箱)もあります。


# セルの出力

セルを実行すると、標準出力と最後の文の結果が、コードのすぐ下の領域に書き出されます。これは「セル出力」と呼ばれています。notebookをファイルメニューから(あるいはctrl+Sやcommand+Sで)保存すると、この出力も一緒に保存されるため、何が起きたかを記録しておくのに便利です。

ツールバーの「Clear All Outputs」をクリックすれば、これをきれいに消去できます。GitHubなどのリポジトリにコードをpushする前には出力をクリアしておくのがよい習慣です。そうしないと、ファイルが大きくなり読みにくくなることがあります。

````python
spams = ["spam"] * 1000
print(spams)

# これを実行した後は出力をクリアしておいたほうがいいかもしれません!
````

# このコースで出会う小さな話題

これは必ずしもnotebook自体の機能というわけではありませんが、notebookで役立つ便利なパッケージなので知っておくとよいでしょう。

`tqdm`パッケージは、任意のイテラブルをラップするだけで、見栄えの良いプログレスバーを表示してくれます。

````python
# プログレスバーがない場合のコードです
# 何が起きているのか気になりながら、10秒ほどかかります

import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
````

````python
# そして今度は、見栄えの良いプログレスバー付きで:

import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
````

````python
# 話は変わりますが、これはmarkdown形式で出力を表示するための便利な方法です

from IPython.display import Markdown, display

display(Markdown("# This is a big heading!\n\n- And this is a bullet-point\n- So is this\n- Me, too!"))
````
