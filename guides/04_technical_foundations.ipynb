私たちが扱う基本的な技術的概念に対して、しっかりと理解できているという安心感を持つことが非常に重要です。それによって、このコース全体の体験がずっと良いものになります。何が起きているのか分からない状態は、とても苦しいものになりかねません。

これらのガイドは、私たちが扱う基盤となる技術についての理解を深めるためのものです。

## トピック1: ChatGPTとOpenAI APIの違い

### OpenAIが提供するChatGPTとGPT APIの違いは何でしょうか?

#### ChatGPTはエンドユーザー向けのツールです。AIを利用する一般消費者向けに設計されたChatプロダクトです。
- 無料プランがあり、より多くの機能を備えた有料のサブスクリプションプランもあります。
- サブスクリプションプランでは、ユーザーはChatプロダクトをほぼ無制限に利用できます。

#### APIは、他の商用プロダクトに取り組むAIエンジニア(ソフトウェアエンジニアやデータサイエンティスト)向けに提供されているサービスです。
- これによって、私たちのような技術者は、「GPT4.1」や「o3」といった基盤モデルにアクセスし、自分たち自身のプロダクトを構築できます。
- やろうと思えば、APIを使って自分たち独自のChatGPT版を作り、エンドユーザーに課金することもできます。
- 多くのAPIと同様に、OpenAIはAPIの利用量に応じてわずかな料金を課します。このコースの多くの例で使用するgpt-4o-miniの場合、API呼び出し1回あたりおよそ0.001ドル程度です。

### 私はChatGPTに月20ドル払っているのに、なぜAPIにさらにお金を払う必要があるのですか?

- これでもう明確になったのではないかと思います。APIは消費者向けではなく、エンジニアが自分たち自身の課金可能なプラットフォームを構築するためのものです。
- もしサブスクリプションを持っているだけでAPIにアクセスできてしまうなら、あなたはChatGPTのようなツールを他の人にもっと安く提供して、OpenAIを廃業させることさえできてしまいます!
- 心に留めておいてください。APIを1回呼び出すだけで、10,000,000,000,000回もの浮動小数点演算が必要になることがあります。その計算には電力が使われているのです!

APIを呼び出す代わりに、オープンソースのモデルを自分のパソコンでローカルに動かすこともできますが、通常はその計算量が1,000分の1程度になります。それでも、ごくわずかとはいえ、その処理はあなたの電気代に影響します。
