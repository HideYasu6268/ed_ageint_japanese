# Vibe codingとデバッグ

「Vibe coding」とは、LLMの助けを借りてコーディングすることを親しみを込めて呼ぶ言葉です。Cursorを使うと特に簡単で楽しく実践できます!ただし、うまく機能させるためのちょっとしたテクニックがあり、それについてはコースの中で紹介しています。ここでは、いくつかの重要なポイントを挙げておきます。

1. ChatGPTやClaudeのようなLLMにコードを書かせる際は、プロンプトに現在の日付を含め、APIの最新バージョンを使うコードを生成するようLLMに指示しましょう。これはOpenAI APIを使う場合に特に重要です!ChatGPTは自分自身のAPIについても、古いバージョンを使おうとすることが頻繁にあります…

2. プロンプトの中で、LLMに対してコードをできるだけ短くシンプルに保つよう依頼しましょう。LLMは不要な余分なチェックをたくさん追加してコードを煩雑にし、問題を見つけにくくしてしまう傾向があります。

3. 複数のLLMに同じ質問をして、最も明確でシンプルな回答を選びましょう。

4. 似たような手法として、1つのLLMから得た回答を別のLLMに渡し、正しいかどうか、あるいはもっとシンプルで分かりやすくできないかを検証してもらいましょう。

5. LLMに、ある問題を解決する方法を3パターン出してもらい、その中で最も分かりやすいものを選びましょう。

## 大規模プロジェクトにおけるVibe Coding

LLMに100〜200行、あるいはそれ以上のコードを一度に生成させるのは避けましょう。すでに十分な知識がある場合を除いて、デバッグして何が問題なのかを把握するのが非常に難しくなります。

代わりに、まず問題を小さく、それぞれ独立してテストできるステップに分解することから始めましょう。問題をどう分解すればいいか分からない場合は、それ自体をLLMに聞いてみるとよいでしょう!

そして、それぞれの構成要素について次のことを行います。
- 上記のコツを使ってLLMにコードを書かせる
- コードをテスト・検証するためのテストもLLMに書かせる
- 自分自身でもテストして、正しく動作していることを確認する

こうすることで、自信を持って大規模なプロジェクトを構築できるようになります。
