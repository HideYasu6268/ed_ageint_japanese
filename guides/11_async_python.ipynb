### では、いくつか例を見てみましょう。

````python
# 非同期関数を定義してみましょう

import asyncio

async def do_some_work():
    print("Starting work")
    await asyncio.sleep(1)
    print("Work complete")
````

````python
# これは何をするでしょうか？

do_some_work()
````

````python
# では、もう一度試してみましょう！

await do_some_work()
````

````python
# これは何がおかしいでしょうか？

async def do_a_lot_of_work():
    do_some_work()
    do_some_work()
    do_some_work()

await do_a_lot_of_work()
````

````python
# 興味深い警告が出ましたね！修正してみましょう

async def do_a_lot_of_work():
    await do_some_work()
    await do_some_work()
    await do_some_work()

await do_a_lot_of_work()
````

````python
# では今度は並行して実行してみましょう
# ここで重要なのは、これはよく知られている「マルチスレッド」とは違うという点です
# asyncioライブラリは単一のスレッド上で動作しますが、あるタスクが待機している間に
# 別のタスクへ切り替えるループの仕組みを使っています

async def do_a_lot_of_work_in_parallel():
    await asyncio.gather(do_some_work(), do_some_work(), do_some_work())

await do_a_lot_of_work_in_parallel()
````

### 最後に - do_a_lot_of_work_in_parallel を呼び出すPythonモジュールを書いてみましょう

上部のリンクを参照してください。モジュール内には次のようなコードが必要になります。

```python
if __name__ == "__main__":
    asyncio.run(do_a_lot_of_work_in_parallel())
```
