### デバッグの公式:10行のコードで再現する

優れたデバッグにはアートとサイエンスの両面があります。習得する一番の方法は実践することです!ですが、ここに要点をまとめます。

1. 再現する
問題を一貫して再現しましょう

2. 単純化する
問題を、再現できる最も単純な形にまで絞り込みましょう。「Edさん、これが動かないので直してください」と言って、100行以上のコードを送ってくる人が今までどれだけいたか、数え切れません。それでは上手くいきません!私が最初に試みるのは、これを核心的な問題まで絞り込むことです。理想的には10〜20行以下のコードにします。ある学生さんには、たった2行のコードで再現できることを示したこともあります!

_実は、これを行うのは、他の誰よりも**あなた自身**にとってずっと簡単なことなのです。_ 地道な作業になることもありますが、これはほぼ常に最良の出発点です。コードを数行まで削り込みましょう。そして、それをやってみると、たいてい問題を自分自身で解決できてしまうものです!

3. 助けを求める
数行のコードで問題を再現できたのに、何が起きているのか分からない場合は、助けを求めましょう!もちろん ChatGPT やその仲間たちに聞いてみる価値はあります。その際は、短い例と完全なスタックトレースを渡してあげてください。

そして私にも連絡してください!私はあなたの学習体験を最高のものにするためにここにいますので、詰まってしまったときは、そこから抜け出せるようお手伝いします。
