### あなたの変更をリポジトリに貢献し、他の人と共有する

皆さんの貢献を取り込めることを、私は大変嬉しく思います。それは他のすべての学生にとって価値のあるものになりますし、私自身もそれを見るのが大好きです!さらに、Github上でリポジトリへのコントリビューターとして認識されるという特典もあります。

PRを作成するための全体的な手順と主要な説明は以下の通りです。
https://edwarddonner.com/pr

提出前に以下を確認してください。
1. PRにはcommunity_contributions内の変更だけが含まれていること(事前に相談した場合を除く)
2. すべてのnotebookの出力がクリアされていること
3. コード全体で2,000行未満であり、ファイル数も多すぎないこと
4. 不要なテストファイルや、冗長すぎるREADME、.env.example、絵文字、その他のLLMが生成した余計な要素を含めないこと!

本当にありがとうございます!
