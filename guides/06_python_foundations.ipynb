## これを踏まえて - PythonのNameErrorを理解する

Pythonを使っているとNameErrorに遭遇することはよくあります。基礎的な知識があれば、NameErrorを常に自信を持ってデバッグし、原因を突き止められるはずです。

notebookで作業している場合の最も一般的な原因は、すべてのセルを順番通りに実行していないため、カーネル上でその何かが定義されていないことです。

NameErrorの直し方が分からない場合は、この[最初のガイド](https://chatgpt.com/share/67958312-ada0-8012-a1d3-62b3a5fcbbfc)と、この[演習付きの2つ目のガイド](https://chatgpt.com/share/67a57e0b-0194-8012-bb50-8ea76c5995b8)を参照し、自信が持てるようになるまで両方に取り組んでみてください。

内容には多少の重複がありますので、すでに自信がある場合は読み飛ばして構いません。

これを終えれば、NameErrorはもう二度と困りものにならないはずです!
