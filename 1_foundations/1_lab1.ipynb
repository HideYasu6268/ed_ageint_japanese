```python
# さて、いよいよ重要なimport文です
# import errorが出た場合は、https://edwarddonner.com/avatar のFAQにあるQ5とQ25を確認してください
# Geminiなど他のLLMプロバイダーを使う場合でも、このOpenAIのimportを使います - 理由はGuide 9を参照してください

from openai import OpenAI
```

```python
# ここで、OpenAIクラスのインスタンスを作成します
# クラスのインスタンスを作成するというのがどういうことか分からない場合は、guidesフォルダ（guide 6）を見てください！
# NameErrorが出た場合は、guidesフォルダ（guide 6）でNameErrorについて学んでください - これは常に即座に修正できます
# OpenAIを使っていない場合は、少し変更するだけで済みます - 詳しい手順はAI APIガイド（guide 9）にあります

openai = OpenAI()
```

```python
# おなじみのOpenAI形式でメッセージのリストを作成します

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# では実際に呼び出してみましょう！何か問題があれば、トラブルシューティングガイドへ
# これは非常に安価なモデルであるGPT 5.4 nanoを使用しています
# APIガイド（guide 9）には、OpenAIよりさらに安価な、あるいは無料の代替手段を使うための具体的な手順が載っています
# NameErrorが出た場合は、guidesフォルダ（guide 6）でNameErrorについて学んでください - これは常に即座に修正できます

response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
print(response.choices[0].message.content)
```

```python
# では、質問を求めてみましょう：

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# 実行してみます - これはGPT 5.4 miniを使用しています。まだ安価ですが、nanoより高性能です

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# 新しいmessagesリストを作成します

messages = [{"role": "user", "content": question}]
messages
```

```python
# モデルに、この難しい質問への回答を求めます

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
answer = response.choices[0].message.content
print(answer)
```
